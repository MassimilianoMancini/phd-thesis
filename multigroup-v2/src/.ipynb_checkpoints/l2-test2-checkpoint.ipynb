```python
from math import sqrt
from statistics import variance
from itertools import combinations
import matplotlib.pyplot as plt

# ---------------------------------------------------------
# 1) Generazione dei valori di v secondo il tuo ciclo
# ---------------------------------------------------------

all_v = []

arr = [0.0, 0.0, 0.0, 0.0]

for i in range(4):
    # prima fase: includi step 0
    if i == 0:
        steps = range(11)          # 0 → 10
    else:
        steps = range(1, 11)       # 1 → 10 (evita duplicati)

    for step in steps:
        arr[i] = step * 0.1
        all_v.append(arr.copy())



# ---------------------------------------------------------
# 2) Calcolo di s, variance, ep per ogni v
# ---------------------------------------------------------

S_values = []
Variance_values = []
EP_values = []

for v in all_v:

    k = len(v)
    d = sqrt(sum(i*i for i in v))

    # ---- caso speciale: vettore nullo ----
    if d == 0:
        S_values.append(0)
        Variance_values.append(0)
        EP_values.append(0)
        continue

    # ---- calcolo di s ----
    s = sum(((i / d) - 1/sqrt(k))**2 for i in v)
    S_values.append(s)

    # ---- calcolo della variance ----
    Variance_values.append(variance(v))

    # ---- calcolo di ep ----
    ep = 0
    j = 0
    for a, b in combinations(v, 2):
        ep += abs(a - b)
        j += 1
    ep /= j
    EP_values.append(ep)


# ---------------------------------------------------------
# 3) Scatter plot con asse X descrittivo
# ---------------------------------------------------------

x = list(range(len(all_v)))

# Creiamo etichette leggibili per ogni v
x_labels = [f"{tuple(round(x,1) for x in v)}" for v in all_v]

plt.figure(figsize=(18, 8))

plt.scatter(x, S_values, label="s", color="blue")
plt.scatter(x, Variance_values, label="variance", color="red")
plt.scatter(x, EP_values, label="ep", color="green")

plt.xticks(x, x_labels, rotation=90, fontsize=8)

plt.xlabel("Vettore v")
plt.ylabel("Valori di s, variance, ep")
plt.title("Andamento di L2, variance ed ep al variare di v")
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```
