```python
from math import sqrt
from statistics import variance
from itertools import combinations
from random import random
```

```python
v = [random() for _ in range(4)]
print(v)
```

```python
k = len(v)
d = 0
for i in v:
    d = d + i**2

d = sqrt(d)
```

```python
s = 0
for i in v:
    s = s + ((i / d) - 1/sqrt(k))**2
print (s)
```

```python
print(variance(v))
```

```python
ep = 0
j = 0
for i in combinations(v, 2):
    ep = ep + abs(i[0] - i[1])
    j = j + 1
ep = ep / j
print (ep)
```
